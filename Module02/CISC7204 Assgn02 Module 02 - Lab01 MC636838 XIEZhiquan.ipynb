# Area Plots<a id="6"></a>


In the last module, we created a line plot that visualized the top 5 countries that contribued the most immigrants to Canada from 1980 to 2013. With a little modification to the code, we can visualize this plot as a cumulative plot, also knows as a **Stacked Line Plot** or **Area plot**.


````python
# 使用排序，令移民去加拿大的总人数从高到低排列。
Canada_immigration.sort_values(['Total'], ascending = False, axis = 0, inplace = True)
# 然后，将移民去加拿大人数最多的前五个国家打印出来：
# 为了避免混淆，存进新的变量，叫做Canada_immigration_top5：
# get the top 5 entries
Canada_immigration_top5 = Canada_immigration.head()
# 现在，这个数据结构应该是国家名在纵列，数据在横行。这是不符合画图的规范的。所以我们要对数据进行转置。
# 将数据进行转置，使用 transpose（）函数，将新的变量命名为 Canada_immigration_transposed：
# transpose the dataframe
Canada_immigration_transposed = Canada_immigration_top5[years].transpose()
# 最后，将其打印出来：
Canada_immigration_transposed.head()
````

Area plots are stacked by default. And to produce a stacked area plot, each column must be either all positive or all negative values (any `NaN`, i.e. not a number, values will default to 0). To produce an unstacked plot, set parameter `stacked` to value `False`.


````python
# let's change the index values of df_top5 to type integer for plotting
# 这里主要是要画出这五个国家移民去加拿大人数的逐年变化图。
# 首先，把转置后表格的索引（年份字符串）转为整数，方便 matplotlib 绘制 X轴年份：
Canada_immigration_transposed.index = Canada_immigration_transposed.index.map(int)
# 绘制面积图 ，kind='area'就是要绘制面积图。stacked=False就是不堆叠，5个国家的区域会重叠在一起。
Canada_immigration_transposed.plot(kind='area', stacked = False, figsize = (20, 10))  # pass a tuple (x, y) size
# 接下来就是设置图表标题，用title函数；设置列标题，用ylabel函数；设置行标题，用xlabel函数。
plt.title('Immigration Trend of Top 5 Countries')
plt.ylabel('Number of Immigrants')
plt.xlabel('Years')
# 最后，打印图像，使用常见的plt.show()函数一气呵成：
plt.show()
````

The unstacked plot has a default transparency (alpha value) at 0.5. We can modify this value by passing in the `alpha` parameter.


````python
# 未堆叠的图表默认透明度(alpha值)为0.5。我们可以通过传递alpha参数来修改此值。
# 注意哦！这里同样要使用转置之后的数据才可以画的出来，否则所有国家都会堆叠在一起。
Canada_immigration_transposed.plot(kind='area', alpha = 0.25, stacked = False, figsize = (20, 10))
# 0 - 1, default value alpha = 0.5
# 图表标题和内部标题的设置和上面的代码完全一致，照抄即可。
plt.title('Immigration Trend of Top 5 Countries')
plt.ylabel('Number of Immigrants')
plt.xlabel('Years')
plt.show()
````

### Two types of plotting

As we discussed in the video lectures, there are two styles/options of plotting with `matplotlib`， plotting using the Artist layer and plotting using the scripting layer.

**Option 1: Scripting layer (procedural method) - using matplotlib.pyplot as 'plt'**

You can use `plt` i.e. `matplotlib.pyplot` and add more elements by calling different methods procedurally; for example, `plt.title(...)` to add title or `plt.xlabel(...)` to add label to the x-axis.

```python
    # Option 1: This is what we have been using so far
    df_top5.plot(kind='area', alpha=0.35, figsize=(20, 10)) 
    plt.title('Immigration trend of top 5 countries')
    plt.ylabel('Number of immigrants')
    plt.xlabel('Years')
```


**Option 2: Artist layer (Object oriented method) - using an `Axes` instance from Matplotlib (preferred)**

You can use an `Axes` instance of your current plot and store it in a variable (eg. `ax`). You can add more elements by calling methods with a little change in syntax (by adding "`set_`" to the previous methods). For example, use `ax.set_title()` instead of `plt.title()` to add title,  or `ax.set_xlabel()` instead of `plt.xlabel()` to add label to the x-axis.

This option sometimes is more transparent and flexible to use for advanced plots (in particular when having multiple plots, as you will see later).

In this course, we will stick to the **scripting layer**, except for some advanced visualizations where we will need to use the **artist layer** to manipulate advanced aspects of the plots.


````python
# option 2: preferred option with more flexibility
# 这里还给出了另一种画图的方式，即axes，令axe同样等于 转置后数据应用.plot()函数
# 然后设置图表标题，绘制图像。
axe = Canada_immigration_transposed.plot(kind='area', alpha=0.35, figsize=(20, 10))
axe.set_title('Immigration Trend of Top 5 Countries')
axe.set_ylabel('Number of Immigrants')
axe.set_xlabel('Years')
axe.plot()
````

# Histograms<a id="8"></a>

A histogram is a way of representing the *frequency* distribution of numeric dataset. The way it works is it partitions the x-axis into *bins*, assigns each data point in our dataset to a bin, and then counts the number of data points that have been assigned to each bin. So the y-axis is the frequency or the number of data points in each bin. Note that we can change the bin size and usually one needs to tweak it so that the distribution is displayed nicely.


**Question:** What is the frequency distribution of the number (population) of new immigrants from the various countries to Canada in 2013?


Before we proceed with creating the histogram plot, let's first examine the data split into intervals. To do this, we will us **Numpy**'s `histrogram` method to get the bin ranges and frequency counts as follows:


````python
# let's quickly view the 2013 data
# 快速获取2013年的示例数据。题目这里加了head，则是直接提取前五行做预览。
# 如果想要预览整个数据，就要把head()给删除。
Canada_immigration['2013'].head()
````

````python
# np.histogram returns 2 values
# 提取2013年移民去加拿大的所有国家数据，然后计算他们的频数、切块边界，以确定数据分布。
count, bin_edges = np.histogram(Canada_immigration['2013'])
print(count) # frequency count
print(bin_edges) # bin ranges, default = 10 bins
````

By default, the `histrogram` method breaks up the dataset into 10 bins. The figure below summarizes the bin ranges and the frequency distribution of immigration in 2013. We can see that in 2013:

*   178 countries contributed between 0 to 3412.9 immigrants
*   11 countries contributed between 3412.9 to 6825.8 immigrants
*   1 country contributed between 6285.8 to 10238.7 immigrants, and so on..

<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/labs/Module%202/images/Mod2Fig1-Histogram.jpg" align="center" width="800">

We can easily graph this distribution by passing `kind=hist` to `plot()`.


````python
# 通过使用 kind=hist，设置画图的类型为histogram，然后将2013年移民去加拿大的人数绘制在图上。
Canada_immigration['2013'].plot( kind = 'hist', figsize = (8, 5))
# 设置图表标题、行标题和列标题：
# add a title to the histogram
plt.title('Histogram of Immigration from 195 Countries in 2013')
# add y-label
plt.ylabel('Number of Countries')
# add x-label
plt.xlabel('Number of Immigrants')
# 最后一步到位展示绘制的图像：
plt.show()
````

In the above plot, the x-axis represents the population range of immigrants in intervals of 3412.9. The y-axis represents the number of countries that contributed to the aforementioned population.

Notice that the x-axis labels do not match with the bin size. This can be fixed by passing in a `xticks` keyword that contains the list of the bin sizes, as follows:


````python
# 'bin_edges' is a list of bin intervals
# 上一张图存在直方图区间边界和X轴刻度对不齐的情况，容易混淆
# 所以先重新算出10个bin的区间边界，然后存进去bin-edges；
count, bin_edges = np.histogram( Canada_immigration['2013'] )
# xticks = bin_edges新增了这个参数，强制把X轴的刻度设置成直方图每个分组的边界值。
Canada_immigration['2013'].plot( kind = 'hist', figsize = (8, 5), xticks = bin_edges)
#最后再执行画图：
plt.title('Histogram of Immigration from 195 countries in 2013') # add a title to the histogram
plt.ylabel('Number of Countries') # add y-label
plt.xlabel('Number of Immigrants') # add x-label
plt.show()
````

*Side Note:* We could use `df_can['2013'].plot.hist()`, instead. In fact, throughout this lesson, using `some_data.plot(kind='type_plot', ...)` is equivalent to `some_data.plot.type_plot(...)`. That is, passing the type of the plot as argument or method behaves the same.

See the *pandas* documentation for more info  [http://pandas.pydata.org/pandas-docs/stable/generated/pandas.Series.plot.html](http://pandas.pydata.org/pandas-docs/stable/generated/pandas.Series.plot.html?utm_medium=Exinfluencer&utm_source=Exinfluencer&utm_content=000026UJ&utm_term=10006555&utm_id=NA-SkillsNetwork-Channel-SkillsNetworkCoursesIBMDeveloperSkillsNetworkDV0101ENSkillsNetwork20297740-2021-01-01).


We can also plot multiple histograms on the same plot. For example, let's try to answer the following questions using a histogram.

**Question**: What is the immigration distribution for Denmark, Norway, and Sweden for years 1980 - 2013?


````python
# let's quickly view the dataset 
# 让我们尝试使用直方图来回答以下问题。
# 问题:丹麦、挪威和瑞典在1980年至2013年期间的移民分布情况如何?
# 首先，使用loc[]方法，提取出丹麦、挪威和瑞典从1980-2013年的移民分布总量。将新变量命名为immigration_of_NorthEuro：
immigration_of_NorthEuro = Canada_immigration.loc[['Denmark', 'Norway', 'Sweden'], years]
# 下表就展示了他们的具体数值：
immigration_of_NorthEuro
````

````python
# generate histogram
Canada_immigration.loc[['Denmark', 'Norway', 'Sweden'], years].plot.hist()
````

That does not look right!

Don't worry, you'll often come across situations like this when creating plots. The solution often lies in how the underlying dataset is structured.

Instead of plotting the population frequency distribution of the population for the 3 countries, *pandas* instead plotted the population frequency distribution for the `years`.

This can be easily fixed by first transposing the dataset, and then plotting as shown below.


````python
# transpose dataframe
# 上面这个图就实在是太丑了。所以我们按照之前的方式，画出histogram：
# 首先，对数据 immigration_of_NorthEuro 进行转置，使用.transpose()函数，将转置完成之后的变量命名为 immigration_of_NorthEuro_transposed：
immigration_of_NorthEuro_transposed = immigration_of_NorthEuro.transpose()
# 然后打印出这个表格的头部，检查：
immigration_of_NorthEuro_transposed.head()
````

````python
# generate histogram
# 下一步，就是绘制直方图histogram。要先设置kind = hist，对immigration_of_NorthEuro_transposed使用.plot()函数：
immigration_of_NorthEuro_transposed.plot(kind='hist', figsize=(10, 6))
# 然后设置title、y轴名字和x轴名字：
plt.title('Histogram of Immigration from Denmark, Norway, and Sweden from 1980 - 2013')
plt.ylabel('Number of Years')
plt.xlabel('Number of Immigrants')
# 绘制出图像：
plt.show()
````

Let's make a few modifications to improve the impact and aesthetics of the previous plot:

*   increase the bin size to 15 by passing in `bins` parameter;
*   set transparency to 60% by passing in `alpha` parameter;
*   label the x-axis by passing in `x-label` parameter;
*   change the colors of the plots by passing in `color` parameter.


````python
# let's get the x-tick values
# 为了进一步提升图的美感和精准度，这里再一次重新计算分段区间：
# 这里将箱子数量增加至15，调整参数：
count, bin_edges_A = np.histogram( immigration_of_NorthEuro_transposed, 15 )
# 然后，将图的透明度alpha设置为60%，将x轴的分段改为新的分段:
# un-stacked histogram 
immigration_of_NorthEuro_transposed.plot( 
    kind ='hist', 
    figsize = (10, 6), 
    bins = 15, alpha = 0.6, 
    xticks = bin_edges_A, 
    color=['coral', 'darkslateblue', 'mediumseagreen'])
# 设置图表标题、y轴标题、x轴标题。
plt.title('Histogram of Immigration from Denmark, Norway, and Sweden from 1980 - 2013')
plt.ylabel('Number of Years')
plt.xlabel('Number of Immigrants')
# 最后，绘制图像：
plt.show()
````

If we do not want the plots to overlap each other, we can stack them using the `stacked` parameter. Let's also adjust the min and max x-axis labels to remove the extra gap on the edges of the plot. We can pass a tuple (min,max) using the `xlim` paramater, as show below.


````python
# 如果我们不希望图块相互重叠，可以使用stacked参数将它们堆叠。我们还可以调整最小和最大x轴标签，以消除图表边缘的多余间隙
# 我们可以使用xlim参数传递一个元组 (min,max)，如下所示。
# - 先算出 15 个分组的区间边界，然后给 X 轴左右两边预留一点空白边距，不让柱子紧贴图片边缘，美观。
count, bin_edges_B = np.histogram( immigration_of_NorthEuro_transposed, 15)
xmin = bin_edges_B[0]  - 10   #  first bin value is 31.0, adding buffer of 10 for aesthetic purposes 
xmax = bin_edges_B[-1] + 10  #  last bin value is 308.0, adding buffer of 10 for aesthetic purposes
# 下一步，画图。因为题目说了要堆叠，所以设置stacked=True，并将x轴的分段设置为新计算的分段；
# stacked Histogram
immigration_of_NorthEuro_transposed.plot( 
    kind='hist',
    figsize=(10, 6), 
    bins=15, xticks = bin_edges_B, 
    color=['coral', 'darkslateblue', 'mediumseagreen'],
    stacked=True, xlim = (xmin, xmax))
# 最后，设置图表标题、行标题和列标题，画图：
plt.title('Histogram of Immigration from Denmark, Norway, and Sweden from 1980 - 2013')
plt.ylabel('Number of Years')
plt.xlabel('Number of Immigrants') 
plt.show()
````

# Bar Charts (Dataframe) <a id="10"></a>

A bar plot is a way of representing data where the *length* of the bars represents the magnitude/size of the feature/variable. Bar graphs usually represent numerical and categorical variables grouped in intervals.

To create a bar plot, we can pass one of two arguments via `kind` parameter in `plot()`:

*   `kind=bar` creates a *vertical* bar plot
*   `kind=barh` creates a *horizontal* bar plot


**Vertical bar plot**

In vertical bar graphs, the x-axis is used for labelling, and the length of bars on the y-axis corresponds to the magnitude of the variable being measured. Vertical bar graphs are particularly useful in analyzing time series data. One disadvantage is that they lack space for text labelling at the foot of each bar.

**Let's start off by analyzing the effect of Iceland's Financial Crisis:**

The 2008 - 2011 Icelandic Financial Crisis was a major economic and political event in Iceland. Relative to the size of its economy, Iceland's systemic banking collapse was the largest experienced by any country in economic history. The crisis led to a severe economic depression in 2008 - 2011 and significant political unrest.

**Question:** Let's compare the number of Icelandic immigrants (country = 'Iceland') to Canada from year 1980 to 2013.


````python
# step 1: get the data
# 同样，对Canada_immigration数据集使用loc方法，提取出Iceland在从1980-2013年移民去加拿大的人数：
# 将新的数据集命名为 immigration_of_Iceand
immigration_of_Iceand = Canada_immigration.loc['Iceland', years]
# 然后打印出数据进行检查：
immigration_of_Iceand.head()
````

````python
# step 2: plot data
# 然后，使用barchart，去展示Iceland移民去加拿大的情况：设置kind=bar；
immigration_of_Iceand.plot( kind='bar', figsize=(10, 6))
# 然后设置图表标题、行标题和列标题：
plt.xlabel('Year') # add to x-label to the plot
plt.ylabel('Number of immigrants') # add y-label to the plot
plt.title('Icelandic immigrants to Canada from 1980 to 2013') # add title to the plot
# 展示图表：
plt.show()
````

The bar plot above shows the total number of immigrants broken down by each year. We can clearly see the impact of the financial crisis; the number of immigrants to Canada started increasing rapidly after 2008.

Let's annotate this on the plot using the `annotate` method of the **scripting layer** or the **pyplot interface**. We will pass in the following parameters:

*   `s`: str, the text of annotation.
*   `xy`: Tuple specifying the (x,y) point to annotate (in this case, end point of arrow).
*   `xytext`: Tuple specifying the (x,y) point to place the text (in this case, start point of arrow).
*   `xycoords`: The coordinate system that xy is given in - 'data' uses the coordinate system of the object being annotated (default).
*   `arrowprops`: Takes a dictionary of properties to draw the arrow:
    *   `arrowstyle`: Specifies the arrow style, `'->'` is standard arrow.
    *   `connectionstyle`: Specifies the connection type. `arc3` is a straight line.
    *   `color`: Specifies color of arrow.
    *   `lw`: Specifies the line width.

I encourage you to read the Matplotlib documentation for more details on annotations:
<https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.annotate.html>.


````python
# 然后我们想将x行的年份标签旋转90度，在括号里加上rot=90；
immigration_of_Iceand.plot(kind='bar', figsize=(10, 6), rot=90)  # rotate the xticks(labelled points on x-axis) by 90 degrees
# 下一步，设置图表标题、行标题和列标题：
plt.xlabel('Year')
plt.ylabel('Number of Immigrants')
plt.title('Icelandic Immigrants to Canada from 1980 to 2013')
# 进一步，为金融危机爆发之后的移民潮加上一个趋势箭头，
# 设置箭头的代码为annotate()函数，分别以xy设置箭头的尖端位置、xytext箭头尾部位置，xycoords坐标系，arrowprobs就是箭头本身的属性。
# Annotate arrow
plt.annotate('',  # s: str. Will leave it blank for no text
             xy=(32, 70),  # place head of the arrow at point (year 2012 , pop 70)
             xytext=(28, 20),  # place base of the arrow at point (year 2008 , pop 20)
             xycoords='data',  # will use the coordinate system of the object being annotated
             arrowprops=dict(arrowstyle='->', connectionstyle='arc3', color='blue', lw=2)
             )
# 然后画出图像：
plt.show()
````

Let's also annotate a text to go over the arrow.  We will pass in the following additional parameters:

*   `rotation`: rotation angle of text in degrees (counter clockwise)
*   `va`: vertical alignment of text \[‘center’ | ‘top’ | ‘bottom’ | ‘baseline’]
*   `ha`: horizontal alignment of text \[‘center’ | ‘right’ | ‘left’]


````python
# 现在，我们再加上一个文本去覆盖箭头：
# 前面的步骤完全相同，直接跳到后面：
immigration_of_Iceand.plot(kind='bar', figsize=(10, 6), rot=90)
plt.xlabel('Year')
plt.ylabel('Number of Immigrants')
plt.title('Icelandic Immigrants to Canada from 1980 to 2013')
# Annotate arrow
# 保留刚刚的箭头设置代码，无需作出改动：
plt.annotate('',  # s: str. Will leave it blank for no text
             xy=(32, 70),  # place head of the arrow at point (year 2012 , pop 70)
             xytext=(28, 20),  # place base of the arrow at point (year 2008 , pop 20)
             xycoords='data',  # will use the coordinate system of the object being annotated
             arrowprops=dict(arrowstyle='->', connectionstyle='arc3', color='blue', lw=2)
             )
# 这里是重点，加上标题。使用annotate函数，第一个要加上文本内容；
# 第二个xy是标题结束的位置，roration是他的倾斜角度；va和ha是文本的垂直对齐方式和水平对齐方式：
# Annotate Text
plt.annotate('2008 - 2011 Financial Crisis',  # text to display
             xy=(28, 30),  # start the text at at point (year 2008 , pop 30)
             rotation=72.5,  # based on trial and error to match the arrow
             va='bottom',  # want the text to be vertically 'bottom' aligned
             ha='left',  # want the text to be horizontally 'left' algned.
             )
# 最后，绘制出图像：
plt.show()
````
