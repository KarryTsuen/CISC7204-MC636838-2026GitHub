# Data Description

The dataset used for this visualization assignment contains *automobile_sales* data representing automobile sales and related variables during recession and non-recession period. 

The dataset includes the following variables:
<br>1. Date: The month end date of sales observation.
<br>2. Recession: A binary variable indicating recession perion; 1 means it was recession, 0 means it was normal.
<br>3. Automobile_Sales: The number of vehicles sold during the period.
<br>4. GDP: The per capita GDP value in USD.
<br>5. Unemployment_Rate: The monthly unemployment rate.
<br>6. Consumer_Confidence: A synthetic index representing consumer confidence, which can impact consumer spending and automobile purchases.
<br>7. Seasonality_Weight: The weight representing the seasonality effect on automobile sales during the period. This variable represents the seasonal effect on automobile sales for a given month. In the automobile industry, sales often fluctuate throughout the year due to seasonal patterns—for example, sales may increase during the festive season or year-end promotions, and decrease during off-peak months like post-holiday winter periods.
A value greater than 1 indicates higher-than-average sales expected for that month due to seasonal trends (e.g., holiday season or new model launches).
A value less than 1 suggests lower-than-average sales due to seasonal slowdowns.
A value around 1 means the season has neutral or average effect on sales.
<br>8. Price: The average vehicle price during the period.
<br>9. Advertising_Expenditure: The advertising expenditure of the company.
<br>10.Vehicle_Type: The type of vehicles sold; Supperminicar, Smallfamiliycar,                 Mediumfamilycar, Executivecar, Sports.
<br>11.Competition: The measure of competition in the market, such as the number of competitors or market share of major manufacturers.
<br>12.Month: Month of the observation extracted from Date..
<br>13.Year: Year of the observation extracted from Date.
<br>
By examining various factors mentioned above from the dataset, you aim to gain insights into how recessions impacted automobile sales for your company.

####  TASK 1.2: How do trends in advertising expenditure correlate with automobile sales during non-recession periods, and what insights can be derived from this relationship.


<details><summary>Click here for a hint</summary>

<p>
    You will require:-
<br>Start by filtering the dataset to include only non-recession periods where the "Recession" column has a value of 0.
<br>Group the data by Year and calculate the average of "Automobile_Sales" and "Advertising_Expenditure".
<br>Ensure you reset the index after aggregation so the DataFrame remains structured properly.
<br>Use sns.lineplot() to plot trends:
<br>One line for average automobile sales (solid green line with circular markers).
<br>Another line for average advertising expenditure (dashed blue line with square markers).
<br>Add labels and a title to make the visualization clear.
<br>Include a legend and a grid for better readability.
</p>
</details>


````python
# 广告支出趋势在非衰退时期与汽车销售有何关联，以及可以从这种关系中得出哪些见解。
# 'Advertising_Expenditure','Automobile_Sales',

# 因为只探究非衰退期的，所以先筛选，把Recession == 0 的行全部拿出来：组成新的数据集 automobile_sales_non_recession:
automobile_sales_non_recession = automobile_sales[automobile_sales['Recession'] == 0]

# 设置图幅大小到合理的范围：
plt.figure( figsize = (12, 6) )

# 关联性分析都是绘制散点图：所以 kind = scatter，x是 广告支出 ，y是非衰退期的汽车销售 ：
# 一般来说，找相关性使用 scatterplot 带上 趋势线 会比较好：
# 然后设置自变量和因变量：
x = automobile_sales_non_recession['Advertising_Expenditure']
y = automobile_sales_non_recession['Automobile_Sales']

# 然后画图！# 因为是 scatterplot，所以kind = scatter：
automobile_sales_non_recession.plot( 
                                    kind = 'scatter', 
                                    x = 'Advertising_Expenditure',
                                    y = 'Automobile_Sales', 
                                    figsize = (10, 6), 
                                    color = 'gold' 
                                    )
# 然后就是设置图表标题、行标题和列标题：
plt.title( 'Advertising Expenditure vs Automobile Sales (Non-recession Period)' )
plt.xlabel( 'Advertising Expenditure' )
plt.ylabel( 'Automobile Sales' )

# 下一步就是绘制刚刚的参考线，这个线遵循一次线性回归：y要写成 y = model_fit[0]+model_fit[1]*x:
# plot line of best fit
# 线性回归拟合，计算model_fit
model_fit = np.polyfit(x, y, 1)
# 绘制拟合直线
plt.plot(x, model_fit[0] * x + model_fit[1], color = 'black')
# 标注公式，位置调整到图内合适地方
plt.annotate('y={0:.2f} x + {1:.2f}'.format(model_fit[0], model_fit[1]), xy=(50, 0.85))
# 打印图表：
plt.show()
# 打印拟合参数
print("slope:", model_fit[0])
print("intercept:", model_fit[1])
````

### TASK 1.3: Use the functionality of **Seaborn Library** to create a visualization to compare the sales trend per vehicle type for a recession period with a non-recession period.


<details><summary>Click here for a hint</summary>

<p>
     To visualize the average number of vehicles sold during recession and non-recession periods, you can use a bar chart
        <br> You will need to group recession average Automobile_Sales and then plot it<br>
    Make use of sns.barplot(x=x,y=y, data = df)</p>
</details>


````python
# 任务1.3:使用Seaborn库的功能创建一个可视化图表，比较衰退期与非衰退期各车型的销售趋势。
# 先按 Year, Vehicle_Type, Recession 分组求销量均值: groupby
# 将新变量命名为 automobile_sales_rece_and_nonrece：
automobile_sales_rece_and_nonrece = automobile_sales.groupby(['Year', 'Vehicle_Type', 'Recession'])['Automobile_Sales'].mean().reset_index()
# 设置图幅大小到合理的范围：
plt.figure( figsize = (14,6) )
# seaborn lineplot，hue区分车型，style区分衰退/非衰退（Recession：1=衰退，0=非衰退）
# 设置图像：
sns.lineplot( 
    data = automobile_sales_rece_and_nonrece,
    x = 'Year',
    y = 'Automobile_Sales',
    hue = 'Vehicle_Type',
    style = 'Recession',
    markers = True
)
# 设置图表标题、x轴标题和y轴标题：
plt.title( 'Sales Trend per Vehicle Type: Recession vs Non-Recession' )
plt.ylabel( 'Average Automobile Sales' )
plt.xlabel( 'Year' )
# 设置图例：
plt.legend( bbox_to_anchor = (1.02, 1), loc = 'upper left' )
plt.tight_layout()
# 一步到位画图：
plt.show()
````

### Now you want to compare the sales of different vehicle types during a recession and a non-recession period
<br>We recommend that you use the functionality of **Seaborn Library** to create this visualization

````python
# 现在你想比较不同车型在经济衰退期和非衰退期的销售情况。
# 我们建议您使用Seaborn库的功能来创建此可视化图表。
# 比较销售情况更适合使用柱状图，所以代码如下：
# 先使用groupby分组统计，衰退期、和非衰退期的各种车型的销量均值；然后将新的变量命名为 automobile_sales_rece_and_nonrece_A:
# groupby()括号里面放的是衰退期'Recession'、各种车型'Vehicle_Type'，括号外面放的是销量 'Automobile_Sales'，后门跟着 reset_index()和mean()两个函数：
automobile_sales_rece_and_nonrece_A = automobile_sales.groupby(['Recession', 'Vehicle_Type'])['Automobile_Sales'].mean().reset_index()
plt.figure(figsize=(10,7))
# 然后，画图。因为画的是柱状图，所以barplot：
sns.barplot(
    data = automobile_sales_rece_and_nonrece_A,
    x = "Vehicle_Type",
    y = "Automobile_Sales",
    hue = "Recession"
)
# 然后设置图表标题、行标题和列标题：
plt.xlabel('Vehicle Type') # add to x-label to the plot
plt.ylabel('Average Automobile Sales') # add y-label to the plot
plt.title('Average Sales of Different Vehicle Types: Recession vs Non-Recession') # add title to the plot
plt.xticks( rotation = 30 )
# 设置图例：
plt.legend(title='Recession (1=Recession,0=Non-recession)', loc="upper right", bbox_to_anchor=(1,1))
# 最后，一步到位，打印出图：
# 展示图表：
plt.tight_layout()
plt.show()
````

 ### TASK 1.6: Use the functionality of Matplotlib to develop a scatter plot to identify the relationship between consumer confidence and automobile sales during recessions.
 #### From the data, develop a scatter plot to identify the relationship between consumer confidence and automobile sales during recession periods.
 <br> Title this plot as 'Consumer Confidence and Automobile Sales during Recessions'


<details><summary>Click here for a hint</summary>

<p>
     You can create dataframe where recession is '1'.
        <br>Pass the 'Consumer_Confidence' and 'Automobile_Sales' to the plt.scatter()</p>

````python
# 任务1.6:使用 Matplotlib 的功能开发一个散点图，以识别消费者信心与汽车销售在经济衰退期间的关系。
# 根据数据，绘制散点图以识别经济衰退期间消费者信心与汽车销售之间的关系。
# 将此图表标题命名为 “ Consumer Confidence and Automobile Sales during Recessions ”：
# 首先，沿用前二题的代码，提取出经济衰退期的数据：
recession_dataset = automobile_sales[automobile_sales['Recession'] == 1]
# 然后，绘制散点图：

# 设置图幅大小到合理的范围：
plt.figure( figsize = (12, 6) )

# 关联性分析都是绘制散点图：所以 kind = scatter，x是 消费者信心 'Consumer_Confidence'，y是 衰退期的汽车销售 'Automobile_Sales' ：
# 一般来说，找相关性使用 scatterplot 带上 趋势线 会比较好：
# 设置自变量和因变量：
x = recession_dataset['Consumer_Confidence']
y = recession_dataset['Automobile_Sales']

# 然后画图！# 因为是 scatterplot，所以kind = scatter：
recession_dataset.plot( 
                        kind = 'scatter', 
                        x = 'Consumer_Confidence',
                        y = 'Automobile_Sales', 
                        figsize = (10, 6), 
                        color = 'red' 
                      )
# 然后就是设置图表标题、行标题和列标题：
plt.title( 'Consumer Confidence and Automobile Sales during Recessions' )
plt.xlabel( 'Consumer Confidence' )
plt.ylabel( 'Automobile Sales' )

# 下一步就是绘制刚刚的参考线，这个线遵循一次线性回归：y要写成 y = model_fit[0]+model_fit[1]*x:
# plot line of best fit
# 线性回归拟合，计算model_fit
model_fit = np.polyfit(x, y, 1)
# 绘制拟合直线
plt.plot(x, model_fit[0] * x + model_fit[1], color = 'black')
# 标注公式，位置调整到图内合适地方
plt.annotate('y={0:.2f} x + {1:.2f}'.format( model_fit[0], model_fit[1]), xy=(50, 0.85) )
# 打印图表：
plt.show()
# 打印拟合参数
print("slope:", model_fit[0])
print("intercept:", model_fit[1])
````
