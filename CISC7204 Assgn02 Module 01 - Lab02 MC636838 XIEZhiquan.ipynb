### Matplotlib.Pyplot

One of the core aspects of Matplotlib is `matplotlib.pyplot`. It is Matplotlib's scripting layer which we studied in details in the videos about Matplotlib. In future labs, we will get to work with the Artist layer as well to experiment first hand how it differs from the scripting layer.

# Line Pots (Series/Dataframe) <a id="4"></a>


**What is a line plot and why use it?**

A line chart or line plot is a type of plot which displays information as a series of data points called 'markers' connected by straight line segments. It is a basic type of chart common in many fields.
Use line plot when you have a continuous data set. These are best suited for trend-based visualizations of data over a period of time.


**Let's start with a case study:**

In 2010, Haiti suffered a catastrophic magnitude 7.0 earthquake. The quake caused widespread devastation and loss of life and about three million people were affected by this natural disaster. As part of Canada's humanitarian effort, the Government of Canada stepped up its effort in accepting refugees from Haiti. We can quickly visualize this effort using a `Line` plot:

**Question:** Plot a line graph of immigration from Haiti using `df.plot()`.


First, we will extract the data series for Haiti.


````python
#Since we converted the years to string, 
#let's declare a variable that will allow us to easily call upon the full range of years:
# 首先，我们要提取出海地的数据，找出海地在1980-2014年移民去加拿大的人数。
years = list(map(str, range(1980, 2014)))
#creating data series
# 对数据集 df_can 使用loc解法，去提取按年份的 海地移民数据。 
# 将提取出来的数据装在新的变量 haiti_can里面，loc[] 中括号里，分别是行名（国家名），和年份的跨度 1980：2014.
haiti_can = df_can.loc['Haiti', 1980:2013] # passing in years 1980 - 2013 to exclude the 'total' column
# 然后将数据打印出来：
haiti_can.head()
# 可以看见数据已经成功提取。
````

Next, we will plot a line plot by appending `.plot()` to the `haiti` dataframe.


````python
# 然后我们对这个新的变量 haiti_can 直接执行 plot()函数，把他的数据绘制出图：
haiti_can.plot()
# 可以看见成功出图，而且在2010年前后移民数量快速从2000余人增长至7000人以上，说明地震促使海地人移民加拿大。
````

*pandas* automatically populated the x-axis with the index values (years), and the y-axis with the column values (population). <br>

Also, let's label the x and y axis using `plt.title()`, `plt.ylabel()`, and `plt.xlabel()` as follows:


````python
# 但是，观察上面的图可以发现，图里缺失了标题、纵列（人数的名字）和横行（年份的名字），所以我们要给他加上去。
haiti.plot(kind='line')
# 首先，给图加上标题，就是对plt执行函数 title(),加标题，括号里就写上标题：Immigrations from Haiti 1980-2014；
plt.title('Immigrations from Haiti 1980-2014')
# 然后，给图加上纵列标签，执行的是 ylabel()函数，即 y轴标签，括号里写上标签名字：Number of immigrants
plt.ylabel('Number of immigrants')
# 然后，给图加上横行标签，执行的是 xlabel() 函数，即 x轴标签，括号里加上标签名字：Years
plt.xlabel('Years')
# 最后，直接对plt执行函数 show() 展示并打印出图：
plt.show()
# need this line to show the updates made to the figure
# 下面这个图像就比上面的美观多了。
````

We can clearly notice how number of immigrants from Haiti spiked up from 2010 as Canada stepped up its efforts to accept refugees from Haiti. Let's annotate this spike in the plot by using the `plt.text()` method.<br>
<br>
However, notice that years are of type *string*. Let's change the type of the index values to *integer* first.


````python
#我们可以清楚地注意到，自2010年以来，随着加拿大加大努力接收来自海地的难民，来自海地的移民人数急剧增加。
# 题目希望使用 plt.text()方法，在图中标注这一峰值。
# 然而，需要注意的是年份是字符串类型。我们先将索引值的类型更改为整数。
haiti.index = haiti.index.map(int) 
# annotate the 2010 Earthquake. 
# 其余的代码和上面的一致，不改：
haiti.plot(kind='line')
plt.title('Immigrations from Haiti 1980-2014')
plt.ylabel('Number of immigrants')
plt.xlabel('Years')
# syntax: plt.text(x, y, label)
# 然后上面的代码给出了 标记数据的固定格式：括号里的前两个元素是xy的值，是用来定位标签加到哪里去的；然后第三个元素是输入文本的内容.
# 我们修改到自己的代码里： plt_new.text( 1995, 6000,'2010 Earthquake in Haiti ->' )
plt.text( 1995, 6000,'2010 Earthquake hit Haiti ->') 
# see note below
# 最后，再次对 plt_new 使用 show() 函数，打印并展示出来：
plt.show() 
````

With just a few lines of code, you were able to quickly identify and visualize the spike in immigration!

Quick note on x and y values in `plt.text(x, y, label)`:
    
     Since the x-axis (years) is type 'integer', we specified x as a year. The y axis (number of immigrants) is type 'integer', so we can just specify the value y = 6000.
    
```python
    plt.text(2000, 6000, '2010 Earthquake') # years stored as type int
```
    If the years were stored as type 'string', we would need to specify x as the index position of the year. Eg 20th index is year 2000 since it is the 20th year with a base year of 1980.
```python
    plt.text(20, 6000, '2010 Earthquake') # years stored as type int
```
    We will cover advanced annotation methods in later modules.

**Question:** Let's compare the number of immigrants from India and China from 1980 to 2013.


Step 1: Get the data set for China and India, and display the dataframe.


````python
### type your answer here
# 题目要我们找出中国和印度的数据集，所以我们直接对数据集 df_can 使用loc方法，中括号里面的代码是 [['India','China'],years]
# 将新的数据集命名为 df_India_and_China :有如下代码：
df_India_and_China = df_can.loc[['India','China'], 1980:2013]
df_India_and_China
# 可以看见数据成功导出：
````

Step 2: Plot graph. We will explicitly specify line plot by passing in `kind` parameter to `plot()`.


````python
### type your answer here
# 题目说要打印出这个数据表，因此，我们直接对数据 df_India_and_China 使用 plot()函数：
df_India_and_China.plot()
````

That doesn't look right...

Recall that *pandas* plots the indices on the x-axis and the columns as individual lines on the y-axis. Since `df_CI` is a dataframe with the `country` as the index and `years` as the columns, we must first transpose the dataframe using `transpose()` method to swap the row and columns.


````python
# 这看起来不对劲...
# 回顾一下，Pandas 将索引绘制在x轴上，将列作为单独的线行和列。我们必须首先使用transpose方法转置该数据框，以交换行条绘制在y轴上。
# 由于 是一个以国家为索引、年份为列的数据框，我们首先要转置这个数据，去把行和列对调一下。使用的是 transpose() 函数；
# 我们用新的 df_India_and_China_transposed
df_India_and_China_transposed = df_India_and_China.transpose()
# 这样转变过来之后，索引值就变成了年份。
````

*pandas* will auomatically graph the two countries on the same graph. Go ahead and plot the new transposed dataframe. Make sure to add a title to the plot and label the axes.


````python
### type your answer here
# 由于上一步转置，导致索引值变成了年份，所以要确保年份是整数，先设置以下代码：
df_India_and_China_transposed.index = df_India_and_China_transposed.index.map(int) 
# 然后，进行绘制。首先是kind = 'line'这个代码：使之转变为 plt 里面的属性
df_India_and_China_transposed.plot( kind ='line')
# 然后，我们给这个图分别加上 标题、y轴标题、x轴标题：
plt.title( 'Immigrants to Canada from China and India' )
plt.ylabel('Number of Immigrants')
plt.xlabel('Years')
# 所有元素设置好，直接用 show() 打印图表：
plt.show()
````

<br>From the above plot, we can observe that the China and India have very similar immigration trends through the years.

**Question:** Compare the trend of top 5 countries that contributed the most to immigration to Canada.


````python
# type your answer here
# 题目要我们画出图，去比较1980-2014年，移民去加拿大人数最多的5个国家的移民变化量趋势。
# 所以，我们要先选出移民最多的5个国家：在这里要新建一个字段叫做 Total：
df_can['Total'] = df_can.loc[:,1980:2013].sum( axis = 1 )
# 排序，取出top5
df_can_top_5 = df_can.sort_values( by = 'Total', ascending = False, axis = 0 )
df_can_top_5 = df_can_top_5.head(5)
df_can_top_5
# 关键改动：只提取年份数字列！
df_top5_year_only = df_can_top_5.loc[:, 1980:2013]
# 转置
df_can_top_5_transposed = df_top5_year_only.transpose()
# 把索引转为整数年份
df_can_top_5_transposed.index = df_can_top_5_transposed.index.map(int)
# 绘图
df_can_top_5_transposed.plot( kind = 'line' )
# 设置图表标题
plt.title("The origin countries of top-5 immigrants to Canada")
# 设置纵列标题:
plt.ylabel("Number of Immigrants")
# 设置横列标题:
plt.xlabel("Years")
# 最后，绘制图表，使用show()函数:
plt.show()
````

### Other Plots

Congratulations! you have learned how to wrangle data with python and create a line plot with Matplotlib. There are many other plotting styles available other than the default Line plot, all of which can be accessed by passing `kind` keyword to `plot()`. The full list of available plots are as follows:

* `bar` for vertical bar plots

* `kde` or `density` for density plots
